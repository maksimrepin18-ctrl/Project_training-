```python
import pandas as pd
import numpy as np

data = {
    'Имя': ['Анна', 'Петр', 'Елена', 'Сергей', 'Мария', 'Иван'],
    'Возраст': [25, 30, 28, '35 лет', 22, np.nan],
    'Город': ['Москва', 'спб', 'Москва', 'Казань', 'спб', 'Москва'],
    'Опыт_работы_лет': [2, 5, 1, 7, 0, np.nan],
    'Продуктивность_%': [105, 98, 110, '95%', 88, 102] 
}

df = pd.DataFrame(data)
print(df)
```

```python
df.info()
```

```python
df.head()
```

```python

df['clean_age'] = df['Возраст'].astype(str)
df['clean_age'] = df['clean_age'].str.replace(r'[^0-9]','', regex=True)
df['clean_age']
```

```python
df['clean_age'] = pd.to_numeric(df['clean_age'], errors='coerce')
df['clean_age']
```

```python
df['clean_age'] = df['clean_age'].fillna(df['clean_age'].mean())
df['clean_age']
```

```python
df['Продуктивность_%']
```

```python

df['clean_progres'] = df['Продуктивность_%'].astype(str)
df['clean_progres']
```

```python
df['clean_progres'] = df['clean_progres'].str.replace(r'[^0-9]','',regex=True)
df['clean_progres']
```

```python
df['clean_city'] = df['Город'].str.strip().str.title()
df['clean_city']
```

```python
df.info()
```

```python
df['clean_experience'] = df['Опыт_работы_лет'].fillna(df['Опыт_работы_лет'].mean())
df['clean_experience']
```

```python
df['clean_experience'] = df['clean_experience'].round(0)
df['clean_experience']
```

```python
df['clean_progres'] = df['clean_progres'].astype(int)
df['clean_progres']
```

```python

df_mask = (df['clean_progres'] > 85) & (df['clean_experience'] < 1)
df_filter = df.loc[df_mask,['Имя', 'Возраст','clean_progres', 'clean_experience']]
df_filter
```
